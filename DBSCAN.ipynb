```python
from sklearn.datasets import make_circles
from sklearn.cluster import KMeans,AgglomerativeClustering,DBSCAN
import matplotlib.pyplot as plt
import scipy.cluster.hierarchy as sch 
```

```python
from sklearn.datasets import make_circles
X, y = make_circles(
    n_samples=800,
    factor=0.4,
    noise=0.05,
    random_state=42
)
plt.scatter(X[:,0],X[:,1])
```

# KMeans performance

```python
wcss = []
for i in range(1,7):
    km = KMeans(n_clusters=i,random_state=42) # default parameters
    km.fit(X)
    wcss.append(km.inertia_)

plt.plot(range(1,7),wcss)
```

KMeans elbow plot gives no information --> no elbow point.

But we know that there should be two clusters --> inner and outer circle so lets test n_clusters=2
However, we can clearly see that wcss at n_clusters = 2 is not an elbow point so most probably KMeans work poor.
But let's see...

```python
kmc = KMeans(n_clusters=2,random_state=42)
y_pred = kmc.fit_predict(X)
plt.scatter(X[:,0],X[:,1],c=y_pred)
```

# Agglomerative hierachial clustering

```python
# dendogram
plt.figure(figsize=(12,5))
fig = sch.dendrogram(sch.linkage(X))
```

From dendogram we can clearly see that the best number of clusters is 2 which is actually true.

```python
for method in ['ward','single','average','complete']:
    ac = AgglomerativeClustering(n_clusters=2,linkage=method)
    y = ac.fit_predict(X)
    plt.figure(figsize=(10,5))
    plt.title(f"{method}")
    plt.scatter(X[:,0],X[:,1],c=y)
```

Single linkage is better because there is a clear boundary in inner and outer circle so the min.distance(single linkage)
method is good here.

# DBSCAN

```python
dbs = DBSCAN(eps=0.3
            ,min_samples=5) 
y2 = dbs.fit_predict(X)

plt.scatter(X[:,0],X[:,1],c=y2)
```

With tuning the epsilon and min_samples we can change the performance by huge amount.
