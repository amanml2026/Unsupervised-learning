```python
from sklearn.datasets import make_circles
from sklearn.cluster import KMeans,AgglomerativeClustering,DBSCAN
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import make_circles
X, y = make_circles(
    n_samples=800,
    factor=0.4,
    noise=0.05,
    random_state=42
)
plt.scatter(X[:,0],X[:,1])
```

# KMeans performance

```python
wcss = []
for i in range(1,7):
    km = KMeans(n_clusters=i,random_state=42) # default parameters
    km.fit(X)
    wcss.append(km.inertia_)

plt.plot(range(1,7),wcss)
```
