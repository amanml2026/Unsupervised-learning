```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from plotly import express as px
```

```python
X,Y = make_blobs(
    n_samples=500,      # Total number of samples
    n_features=2,       # Two features
    centers=5,          # Number of clusters
    cluster_std=1.0,    # Standard deviation of each cluster
    random_state=42
)
```

```python
plt.scatter(X[:,0],X[:,1])
```

```python
# elbow method
wcss=[]
for i in range(1,11):
    km = KMeans(n_clusters=i)
    km.fit_predict(X)
    wcss.append(km.inertia_)
print(wcss)
plt.plot(range(1,11),wcss)
```

```python
km = KMeans(n_clusters=5)
y_pred = km.fit_predict(X)

plt.scatter(X[:,0],X[:,1],c=y_pred)
```

# 3-D clustering

```python
# 3-d dataset
x,y= make_blobs(
    n_samples=500,      # Total number of samples
    n_features=3,       # Two features
    centers=4,          # Number of clusters
    cluster_std=1.0,    # Standard deviation of each cluster
    random_state=42
)
```

```python
fig = px.scatter_3d(x=x[:,0],y=x[:,1],z=x[:,2])
fig.update_traces(
    marker=dict(
        line=dict(
            width=2,          # Thickness of the borderline
            color='black'     # Color of the borderline
        )
    )
)

fig.show()
```

```python
# elbow method
wcss=[]
for i in range(1,11):
    km = KMeans(n_clusters=i)
    km.fit_predict(x)
    wcss.append(km.inertia_)
print(wcss)
plt.plot(range(1,11),wcss)
```

```python
kmc = KMeans(n_clusters=4)
y_pred_3d = kmc.fit_predict(x)
```

```python
# visualization of clusters
fig = px.scatter_3d(x=x[:,0],y=x[:,1],z=x[:,2],color=y_pred_3d)
fig.update_traces(
    marker=dict(
        line=dict(
            width=2,          # Thickness of the borderline
            color='black'     # Color of the borderline
        )
    )
)

fig.show()
```
